## Data Pre-processing

### encoding

```python
# Mengubah data object menjadi numerik
df['school'] = df['school'].map({'GP': 0, 'MS': 1})
df['sex'] = df['sex'].map({'F': 0, 'M': 1})
df['address'] = df['address'].map({'U': 0, 'R': 1})
df['famsize'] = df['famsize'].map({'LE3': 0, 'GT3': 1})
df['Pstatus'] = df['Pstatus'].map({'A': 0, 'T': 1})
```

```python
df.select_dtypes(include='object').nunique()
```

```python
# ubah jadi biner
binary_map = {'no': 0, 'yes': 1}
cols_binary = ['schoolsup', 'famsup', 'paid', 'activities', 'nursery', 'higher', 'internet', 'romantic']

for col in cols_binary:
    df[col] = df[col].map(binary_map)
```

```python
# One-hot encoding untuk kategori>3
df = pd.get_dummies(df, columns=['Mjob', 'Fjob', 'reason', 'guardian'], drop_first=True)
```

## PCA

```python
from sklearn.decomposition import PCA # Import PCA here
import numpy as np
import matplotlib.pyplot as plt

pca = PCA()
pca.fit(df_scaled)

# Ambil nilai-nilai PCA
eigenvalues = pca.explained_variance_
explained_var_ratio = pca.explained_variance_ratio_
cumulative_var_ratio = np.cumsum(explained_var_ratio)

# Buat DataFrame mirip SPSS Total Variance Explained
variance_df = pd.DataFrame({
    'Komponen': [f'PC{i+1}' for i in range(len(eigenvalues))],
    'Eigenvalue': eigenvalues,
    'Rasio Varians (%)': explained_var_ratio * 100,
    'Varians Kumulatif (%)': cumulative_var_ratio * 100
})

# Tampilkan semua
print("=== Total Variance Explained (Semua Komponen) ===")
print(variance_df.round(2))

# Filter hanya komponen dengan Eigenvalue > 1
eigen_df_over1 = variance_df[variance_df['Eigenvalue'] > 1]

print("\n=== Komponen dengan Eigenvalue > 1 ===")
print(eigen_df_over1.round(2))
```

```python
plt.figure(figsize=(12, 6))
plt.bar(variance_df['Komponen'], variance_df['Rasio Varians (%)'], color='teal')
plt.axhline(y=5, color='red', linestyle='--', label='Batas 5%')
plt.title('Barplot Rasio Varians per Komponen (Eigenvalue)')
plt.xlabel('Komponen Utama')
plt.ylabel('Rasio Varians (%)')
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Scree Plot
plt.figure(figsize=(10, 6))
plt.plot(np.arange(1, len(pca.explained_variance_)+1), pca.explained_variance_, marker='o', linestyle='-')
plt.title('Scree Plot - Nilai Eigen per Komponen Utama')
plt.xlabel('Komponen Utama (PC)')
plt.ylabel('Nilai Eigen')
plt.axhline(y=1, color='red', linestyle='--', label='Batas Eigenvalue = 1')
plt.xticks(np.arange(1, len(pca.explained_variance_)+1))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Komunalitas dan Component Matrix

```python
# Communalities
# pca sudah di-fit dan kita pakai 17 PC
loadings = pca.components_[:17]  # 17 x n_features
# Transpose dulu biar kolom = PC, baris = variabel
loadings_df = pd.DataFrame(loadings.T, index=df_scaled.columns, columns=[f'PC{i+1}' for i in range(17)])

# Hitung communalities = jumlah kuadrat loading tiap variabel
communalities = (loadings_df ** 2).sum(axis=1).sort_values(ascending=False)

# Tampilkan sebagai DataFrame
communalities_df = communalities.reset_index()
communalities_df.columns = ['Variabel', 'Komunalitas']

# Lihat hasil top-n
print(communalities_df.head(10))
```

```python
# Jumlah komponen utama dengan eigenvalue > 1
n_komponen_eigen_gt1 = 17  # ganti sesuai jumlah PC dengan eigenvalue > 1

loadings = pd.DataFrame(
    pca.components_[:n_komponen_eigen_gt1],  # Ambil hanya komponen dengan eigenvalue > 1
    columns=df_scaled.columns,
    index=[f'PC{i+1}' for i in range(n_komponen_eigen_gt1)]
)

# Tampilkan 5 besar variabel kontributor untuk setiap PC
for i in range(n_komponen_eigen_gt1):
    pc = f'PC{i+1}'
    print(f'\nTop 5 variabel kontributor untuk {pc}:')
    top_vars = loadings.loc[pc].abs().sort_values(ascending=False).head(5)
    print(top_vars)
```

```python
# Hitung loading matrix
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)

# Buat DataFrame loading matrix dengan nama variabel dan komponen
# Use X_scaled.columns as the index because PCA was performed on X_scaled
df_loadings = pd.DataFrame(loadings,
                           index=df_scaled.columns,
                           columns=[f'Component {i+1}' for i in range(loadings.shape[1])])

print("Component Matrix (Loading Matrix):")
print(df_loadings)
```
